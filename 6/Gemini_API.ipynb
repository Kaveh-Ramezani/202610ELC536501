# 2. Authentication

```python
from google import genai
from google.colab import userdata

# Securely fetch the API key
try:
    api_key = userdata.get('GEMINI_API_KEY')
    print("API Key successfully retrieved from Colab Secrets.")
except Exception as e:
    print("Error: Could not find 'GEMINI_API_KEY' in Colab Secrets.")

# Initialize the modern client
client = genai.Client(api_key=api_key)
```

# 3. API Request

```python
from google.genai import types

# Define your prompt
prompt = "Briefly explain the physical mechanism of using a primary coil to transmit power to a secondary coil located inside an implanted neural stimulator chip."

print("Sending request to Gemini API...\n")

# Generate the response using the current 2.5 Flash model
response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=prompt,
    config=types.GenerateContentConfig(
        system_instruction="You are an expert engineering assistant specializing in biomedical electronics.",
        temperature=0.7,
        max_output_tokens=1024 # Increased from 150 to give it plenty of room to talk!
    )
)

print("--- API Response ---")
print(response.text)
```
